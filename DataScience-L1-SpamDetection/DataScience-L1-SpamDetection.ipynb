# Email Spam Detection with Machine Learning
Objective: classify spam and ham messages using text preprocessing, TF-IDF, Naive Bayes, and a second classifier.


```python
import re
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,classification_report,confusion_matrix
url="https://raw.githubusercontent.com/justmarkham/DAT8/master/data/sms.tsv"
df=pd.read_csv(url,sep="\t",header=None,names=["label","text"]); df["clean_text"]=df.text.str.lower().str.replace(r"[^a-z0-9\s]"," ",regex=True).str.replace(r"\s+"," ",regex=True).str.strip()
print(df.shape); print(df.label.value_counts()); print((df.label.value_counts(normalize=True)*100).round(2))
```

TF-IDF measures how important a word is to a document relative to the full collection: frequent document-specific words receive higher weight than words common to every message. Recall matters in spam detection because missed spam messages reach the user.


```python
Xtr,Xte,ytr,yte=train_test_split(df.clean_text,df.label,test_size=.2,stratify=df.label,random_state=42)
models={"Multinomial Naive Bayes":MultinomialNB(),"Logistic Regression":LogisticRegression(max_iter=1000)}
for name,est in models.items():
 pipe=Pipeline([("tfidf",TfidfVectorizer(stop_words="english",ngram_range=(1,2))),("model",est)]); pipe.fit(Xtr,ytr); pred=pipe.predict(Xte)
 print("\n",name); print("accuracy",accuracy_score(yte,pred),"precision",precision_score(yte,pred,pos_label="spam"),"recall",recall_score(yte,pred,pos_label="spam"),"f1",f1_score(yte,pred,pos_label="spam")); print(classification_report(yte,pred)); sns.heatmap(confusion_matrix(yte,pred),annot=True,fmt="d",cmap="Blues",xticklabels=["ham","spam"],yticklabels=["ham","spam"]); plt.title(name); plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.show()
```
