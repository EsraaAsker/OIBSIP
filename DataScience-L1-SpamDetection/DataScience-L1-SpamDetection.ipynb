# SMS Spam Detection

Classify SMS messages as spam or ham using TF-IDF and two complementary classifiers.

**Portfolio project:** reproducible analysis with documented assumptions, visual evidence, and model evaluation.

```python
import warnings
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

## Dataset and quality checks
UCI SMS Spam Collection mirror: https://raw.githubusercontent.com/justmarkham/DAT8/master/data/sms.tsv

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,ConfusionMatrixDisplay
url="https://raw.githubusercontent.com/justmarkham/DAT8/master/data/sms.tsv"
df=pd.read_csv(url,sep="\t",header=None,names=["label","text"]); print(df.shape,df.label.value_counts())
Xtr,Xte,ytr,yte=train_test_split(df.text,df.label,test_size=.2,stratify=df.label,random_state=42)
for name,clf in {"Naive Bayes":MultinomialNB(),"Logistic Regression":LogisticRegression(max_iter=1000)}.items():
 pipe=Pipeline([("tfidf",TfidfVectorizer(lowercase=True,stop_words="english",ngram_range=(1,2))),("clf",clf)]); pipe.fit(Xtr,ytr); pred=pipe.predict(Xte); print(name); print(classification_report(yte,pred)); ConfusionMatrixDisplay.from_predictions(yte,pred); plt.title(name); plt.show()
```
