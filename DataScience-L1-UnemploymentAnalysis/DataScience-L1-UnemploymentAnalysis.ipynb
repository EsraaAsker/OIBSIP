# Unemployment Analysis with Python
Objective: uncover regional and temporal unemployment trends in India and compare pre-COVID and post-COVID periods.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")
df=pd.read_csv("data/unemployment.csv")
df.columns=df.columns.str.strip()
date_col="Date"; rate_col="Estimated Unemployment Rate (%)"; emp_col="Estimated Employed"; labour_col="Estimated Labour Participation Rate (%)"
df[date_col]=pd.to_datetime(df[date_col],errors="coerce")
for c in [rate_col,emp_col,labour_col]: df[c]=pd.to_numeric(df[c],errors="coerce")
print("shape:",df.shape); print(df.dtypes); print("nulls:",df.isnull().sum()); display(df.describe(include="all"))
```

```python
regional=df.groupby("Region",as_index=False)[rate_col].mean().sort_values(rate_col,ascending=False)
month=df.groupby(date_col,as_index=False)[rate_col].mean()
plt.figure(figsize=(12,5))
for region in regional.head(3).Region:
 x=df[df.Region==region].sort_values(date_col); plt.plot(x[date_col],x[rate_col],marker="o",label=region)
plt.title("Unemployment rate over time: three major regions"); plt.legend(); plt.show()
regional.head(10).plot.barh(x="Region",y=rate_col,figsize=(8,5),legend=False,title="Top 10 regions by average unemployment"); plt.gca().invert_yaxis(); plt.show()
```

```python
corr=df[[rate_col,emp_col,labour_col]].corr()
sns.heatmap(corr,annot=True,cmap="vlag",center=0); plt.title("Correlation heatmap"); plt.show()
cut=pd.Timestamp("2020-03-01")
pre=df[df[date_col]<cut][rate_col].mean(); post=df[df[date_col]>=cut][rate_col].mean()
print({"pre_covid_mean":pre,"post_covid_mean":post,"difference":post-pre})
print("Observation: compare the two means and the time-series shape; avoid treating correlation as causation.")
```
